# Load and test the OSTech laser controller
Below some of the basic functionality of the OSTech laser controller is demonstrated.
In particular the functions for the following are demonstrated:
 - The controller is connected
 - Output is turned on and off
 -  Current is changed
 - The controller is disconnected

```python
from OSTech_laser_controller import OSTechLaserController
```

```python
controller = OSTechLaserController()
port = "COM12" # Change here to the port of your laser controller
controller.connect(port)
controller.init_laser()
```

## Toggle output on

```python
print("Output on: ", controller.is_output_on())
controller.turn_on_output()
print("Output on: ", controller.is_output_on())
```

## Turn output off again

```python
controller.turn_off_output()
print("Output on: ", controller.is_output_on())
```

## Change the current on the laser 

```python
controller.set_current(123) # Sets the current of the controller to 123 mA
```

## Disconnect the controller to let other software use it

```python
controller.disconnect()
```
